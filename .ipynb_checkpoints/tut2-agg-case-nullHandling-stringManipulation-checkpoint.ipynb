```python
%%sql

SELECT customer_name, salary,
CASE
    WHEN salary > 50000 THEN 'High'
    WHEN salary BETWEEN 20000 AND 50000 THEN 'Medium'
    WHEN salary <= 20000 THEN 'Low'
    ELSE 'No Data'
END AS SPENDER
FROM sql_tut.customer;
```

```python
%%sql

SELECT customer_name,
CASE
    WHEN phone_num IS NULL AND email_address IS NULL THEN '9999'
    WHEN phone_num IS NULL THEN email_address
    ELSE phone_num
END AS CONTACT
FROM sql_tut.customer;
```

## NULL HANDLING
    -> the value of column can be null type of it is hard coded as string 
    -> if it is of type null in where condition use "IS NULL" otherwise it will not show
    -> Null can be handled by 2 methods
        - COALESCE() -- more than 2 parameters
        - IFNULL() -- only 2 parameters

```python
%%sql

SELECT * FROM sql_tut.customer
```

```python
%%sql 
SELECT 
    customer_name,
    COALESCE(salary, 0.00)
FROM sql_tut.customer
```

```python
%%sql 
SELECT 
    customer_name,
    IFNULL(salary, 0.00)
FROM sql_tut.customer
```

## String Manupulation

```python
%%sql
DROP TABLE IF EXISTS sql_tut.cust;
CREATE TABLE sql_tut.cust(
    id INT PRIMARY KEY,
    customer_name VARCHAR(100),
    phone_num VARCHAR(100),
    email_address VARCHAR(100),
    address VARCHAR(220)
);
```

```python
%%sql
SELECT 
LENGTH(customer_name) as name_length,
UPPER (CONCAT (customer_name,'-',phone_num)) as name_p,
SUBSTRING(customer_name, 1, 5) as cust_subs,
TRIM('   Chennai   ') AS trimmed_city,
RPAD (customer_name, 15, '*') AS right_pad,
REPLACE(customer_name, ' ', '_') AS update_name,
INSTR (customer_name, 'a') AS position_of_a,
RIGHT (customer_name, 5) AS five_charac,
REVERSE (customer_name) AS reverse
FROM sql_tut.cust;
```

```python
%%sql
SELECT 
    customer_name,
    (SELECT MAX(amount) FROM sql_tut.orders) AS max_amount
FROM sql_tut.customers;
```

```python
%%sql
SELECT customer_name
FROM sql_tut.customers
WHERE customer_id IN (
    SELECT DISTINCT customer_id
    FROM sql_tut.orders
)
```

### DISTINCT 
    -> It returns only the unique value

### EXISTS
    -> If the the value exists it returns all the rows else return nothing

### CURDATE()
    -> Returns the current date in 'YYYY-MM-DD' string format

```python
%%sql
SELECT customer_name
FROM sql_tut.customers
WHERE EXISTS (
    SELECT * from sql_tut.customers where city = 'Banglore'
)
```

```python
%%sql
SELECT
customer_name,
(
    SELECT SUM(amount) 
    FROM sql_tut.orders o
    WHERE o.customer_id = c.customer_id
) AS total_orders
FROM sql_tut.customers c;
```

```python
%%sql
SELECT
    name_length,
    name_p,
    cust_subs,
    trimmed_city,
    right_pad,
    update_name,
    position_of_a,
    five_charac,
    reverse
FROM 
(SELECT 
    id,
    LENGTH(customer_name) as name_length,
    UPPER (CONCAT (customer_name,'-',phone_num)) as name_p,
    SUBSTRING(customer_name, 1, 5) as cust_subs,
    TRIM('   Chennai   ') AS trimmed_city,
    RPAD (customer_name, 15, '*') AS right_pad,
    REPLACE(customer_name, ' ', '_') AS update_name,
    INSTR (customer_name, 'a') AS position_of_a,
    RIGHT (customer_name, 5) AS five_charac,
    REVERSE (customer_name) AS reverse
FROM sql_tut.cust)
AS subquery;
```

```python
%%sql
SELECT
    customer_name,
    CASE
        WHEN (SELECT SUM(amount) FROM sql_tut.orders o WHERE o.customer_id = c.customer_id) >
             (SELECT AVG(amount) FROM sql_tut.orders) THEN 'Above Average'
        ELSE 'Below Average'
    END as order_category
FROM sql_tut.customers c
```
