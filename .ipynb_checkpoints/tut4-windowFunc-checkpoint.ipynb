## WINDOW FUNCTIONS
    -> it performs calculation across a set of rows related to the current row
    -> without changing the num of rows in result
    -> it contains 'OVER' keyword

```python
%%sql

SHOW DATABASES;
```

```python
%%sql
USE sql_tut
```

```python
%%sql
CREATE TABLE sales (
    TransactionId INT,
    Store VARCHAR(50),
    SalesAmount DECIMAL(10,2)
)
```

```python
%%sql

INSERT INTO Sales (TransactionId, Store, SalesAmount)
VALUES
    (1, 'A', 100),
    (2, 'A', 150),
    (3, 'B', 300),
    (4, 'A', 200),
    (5, 'B', 250);
```

```python
%%sql
SELECT * FROM sales
```

```python
%%sql

SELECT *,
       RANK() OVER (ORDER BY SalesAmount ASc) AS rowNum
FROM sales;
```

```python
%%sql
TRUNCATE sales
```

```python
%%sql
SELECT TransactionId, Store, SalesAmount, SUM(SalesAmount)
OVER (PARTITION BY Store) AS TotalSales
FROM sales;
```

```python
%%sql
SELECT TransactionId, Store, SalesAmount, ROW_NUMBER()
OVER (PARTITION BY Store ORDER BY SalesAmount DESC) AS RowNum
FROM sales;
```

```python
%%sql
DROP TABLE IF EXISTS employees;
CREATE TABLE employees (
    EmployeeId INT,
    Name VARCHAR(50),
    Department VARCHAR(50),
    HireDate DATE
)
```

```python
%%sql
INSERT INTO employees (EmployeeId, Name, Department, HireDate)
VALUES
    (1, 'Alice', 'HR', '2020-05-01'),
    (1, 'Alice', 'HR', '2022-06-01'),
    (2, 'Bob', 'IT', '2021-07-15'),
    (3, 'Charlie', 'Finance', '2020-01-30'),
    (3, 'Charlie', 'Finance', '2022-05-26');
```

```python
%%sql
SELECT * FROM employees
```

```python
%%sql
SELECT *, ROW_NUMBER() 
OVER (PARTITION BY Department ORDER BY HireDate ASC) AS RowNum
FROM employees
```

```python
%%sql
WITH EmployeeRow AS (
    SELECT *, ROW_NUMBER() 
    OVER (PARTITION BY Department ORDER BY HireDate ASC) AS RowNum
    FROM employees
)
SELECT * FROM EmployeeRow WHERE RowNum =1;
```

```python
%%sql
drop table if exists students;
CREATE TABLE Students (
    StudentID INT,
    StudentName VARCHAR(100),
    Examscore INT
)
```

```python
%%sql
INSERT INTO Students (StudentID, StudentName, Examscore)
VALUES
    (1, 'Alice', 95),
    (2, 'Bob', 82),
    (3, 'Charlie', 82),
    (4, 'David', 67),
    (5, 'Eva', 68);
```

```python
%%sql
SELECT * FROM Students
```

```python
%%sql
SELECT *, RANK() 
    OVER (ORDER BY Examscore DESC) AS RankScore
    FROM Students
```

```python
%%sql
TRUNCATE Students
```

```python
%%sql
SELECT *, DENSE_RANK() 
    OVER (ORDER BY Examscore DESC) AS RankScore
    FROM Students
```
