```python
%%sql
CREATE TABLE employees (
    emp_id INT PRIMARY KEY,
    name VARCHAR(220) NOT NULL,
    hire_date DATE NOT NULL,
    salary DECIMAL(10,2)
)
```

## CTAS 
    -> to create a table from the exsisting table

```python
%%sql

CREATE TABLE high_paid_employees AS
SELECT
    emp_id,
    name,
    salary
FROM employees
WHERE salary > 60000;
```

## TEMPORARY TABLE
    -> to create a table to store data for that session
    -> after that session it will be deleted

```python
%%sql

CREATE TEMPORARY TABLE (
    emp_id INT,
    salary
);
INSERT INTO temp_high_paid_employees
SELECT
    employee_id,
    salary
FROM employees
WHERE salary > 60000,
```
