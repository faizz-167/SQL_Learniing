## VIEW
    -> Views do not physically store data. Instead, they derive their data from underlying tables whenever they are queried.
    -> if i add some data to the the orginal table the view automatically updates

```python
%%sql
CREATE TABLE sql_tut.employees (
    emp_id INT PRIMARY KEY,
    first_name VARCHAR(100),
    last_name VARCHAR(100),
    email VARCHAR(100),
    hire_date DATE,
    job_title VARCHAR(100),
    salary DECIMAL(10,2)
);
```

```python
%%sql
SELECT
r.name as res_name,
o.date
FROM sql_tut.restaurants r
LEFT JOIN sql_tut.res_orders o
ON r.id = o.restaurant_id
```

### FULL OUTER JOIN
    -> There is no full outer join in my sql
    -> This can achieve the same result by combining a LEFT JOIN and a RIGHT JOIN using the UNION operato

```python
%%sql
SELECT
    r.name as res_name,
    o.date
FROM
    sql_tut.restaurants r
LEFT JOIN
    sql_tut.res_orders o ON r.id = o.restaurant_id
UNION
SELECT
    r.name as res_name,
    o.date
FROM
    sql_tut.restaurants r
RIGHT JOIN
    sql_tut.res_orders o ON r.id = o.restaurant_id
```
